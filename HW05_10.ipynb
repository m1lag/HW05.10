Number 1

```python
temperatures = [2, 5, 4, 6, 7, 3, 1]

avg_temp = sum(temperatures) / len(temperatures)
max_temp = max(temperatures)
min_temp = min(temperatures)
difference = max_temp - min_temp
print(f" Avg  {avg_temp:.2f}")
print(f" Max {max_temp}")
print(f" Min {min_temp}")
print(f" difference {difference}")
```

Number 2

```python
expenses = [1200, 1500, 800, 600, 1000, 400, 900]

total = sum(expenses)
average = total / len(expenses)
maximum = max(expenses)
minimum = min(expenses)

days_over_1000 = sum(1 for x in expenses if x > 1000)

print(f" Total {total}")
print(f" Avr {average:.2f}")
print(f" Max {maximum}")
print(f" Min {minimum}")
print(f" days_over {days_over_1000}")
```

Number 3

```python
from collections import Counter

votes = [
    "apple",
    "banana",
    "apple",
    "orange",
    "banana",
    "apple",
    "apple",
    "apple",
    "orange"
]

counter = Counter(votes)

print("Fruits")

for fruit, count in counter.items():
    print(f"{fruit} {count}")

leader = counter.most_common(1)[0]

print("\nleader:")
print(leader)

print("\nTop 3:")
for fruit, count in counter.most_common(3):
    print(f"{fruit}: {count}")
```

Number 4

```python
math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]

subjects = {
    "Math": math,
    "Physics": physics,
    "English": english
}

averages = {}

for subject, grades in subjects.items():
    avg = sum(grades) / len(grades)
    mn = min(grades)
    mx = max(grades)

    averages[subject] = avg

    print(f"{subject} {avg:.2f} {mn} {mx}")

best_subject = max(averages, key=averages.get)

print("\nbest_subject:")
print(best_subject)
```
